## Aggregate data per user

[Now, as the data is clean, aggregate data per user per period in order to have just one record per user per period. It should ease the further analysis a lot.]

```python
# Calculate the number of calls made by each user per month. Save the result.
calls_per_month = df_calls.groupby(['user_id', 'month']).agg(
    calls_count=('id', 'count')
).reset_index()

# View result
calls_per_month.head()
```

```python
# Calculate the amount of minutes spent by each user per month. Save the result.
# Calculate total minutes spent by each user per month
minutes_per_month = df_calls.groupby(['user_id', 'month']).agg(
    total_minutes=('duration', 'sum')
).reset_index()

# View result
minutes_per_month.head()
```

```python
# Calculate the number of messages sent by each user per month. Save the result.
# Calculate number of messages per user per month
messages_per_month = df_messages.groupby(['user_id', 'month']).agg(
    messages_count=('id', 'count')
).reset_index()

# View result
messages_per_month.head()
```

```python
# Calculate the volume of internet traffic used by each user per month. Save the result.
# Calculate total internet traffic per user per month (in MB)
internet_per_month = df_internet.groupby(['user_id', 'month']).agg(
    total_mb=('mb_used', 'sum')
).reset_index()

# View result
internet_per_month.head()
```

## Test statistical hypotheses

[Test the hypothesis that the average revenue from users of the Ultimate and Surf calling plans differs.]

[Formulate the null and the alternative hypotheses, choose the statistical test, decide on the alpha value.]

**Null hypothesis (H₀):** The average revenue from Surf and Ultimate users is the same.

**Alternative hypothesis (H₁):** The average revenue from Surf and Ultimate users is different.

Significance level: α = 0.05.

### Hypotheses: Surf vs Ultimate revenue

- Null hypothesis (H₀): The average revenue from Surf and Ultimate users is the same.
- Alternative hypothesis (H₁): The average revenue from Surf and Ultimate users is different.

- Significance level: α = 0.05
- Statistical test: Independent two-sample t-test (Welch’s t-test), because we compare the means of two independent groups.

```python
from scipy import stats

alpha = 0.05

# IMPORTANT: remove missing values first
df_agg = df_agg.dropna(subset=['revenue'])

surf_revenue = df_agg[df_agg['plan'] == 'surf']['revenue']
ultimate_revenue = df_agg[df_agg['plan'] == 'ultimate']['revenue']

t_stat, p_value = stats.ttest_ind(
    surf_revenue,
    ultimate_revenue,
    equal_var=False
)

print("p-value:", p_value)

if p_value < alpha:
    print("Reject H₀: revenue differs between plans.")
else:
    print("Fail to reject H₀: no significant difference in revenue.")
```

### Hypotheses: NY-NJ vs other regions revenue

We test whether the average revenue differs between users in the NY-NJ area and users in other regions.

- Null hypothesis (H₀): The average revenue from users in the NY-NJ area is the same as from users in other regions.
- Alternative hypothesis (H₁): The average revenue from users in the NY-NJ area is different from other regions.

- Significance level: α = 0.05
- Statistical test: Independent two-sample t-test (Welch’s t-test), because we compare the means of two independent groups.

[Formulate the null and the alternative hypotheses, choose the statistical test, decide on the alpha value.]

```python
ny_nj_revenue = df_agg[df_agg['city'].str.contains('NY-NJ')]['revenue']
other_revenue = df_agg[~df_agg['city'].str.contains('NY-NJ')]['revenue']

ny_nj_revenue = ny_nj_revenue.dropna()
other_revenue = other_revenue.dropna()

from scipy import stats

alpha = 0.05

t_stat, p_value = stats.ttest_ind(
    ny_nj_revenue,
    other_revenue,
    equal_var=False
)

print("p-value:", p_value)

if p_value < alpha:
    print("Reject H₀: revenue differs between regions.")
else:
    print("Fail to reject H₀: no significant difference in revenue.")
```

# Test the hypotheses
## Hypothesis Test

**Null hypothesis (H₀):** The average revenue from users in the NY-NJ area is the same as the average revenue from users in other regions.

**Alternative hypothesis (H₁):** The average revenue from users in the NY-NJ area is different from the average revenue from users in other regions.

**Significance level:** α = 0.05


```python
# Test the hypotheses
from scipy import stats

alpha = 0.05

# Split groups
ny_nj_revenue = df_agg[df_agg['city'].str.contains('NY-NJ')]['revenue']
other_revenue = df_agg[~df_agg['city'].str.contains('NY-NJ')]['revenue']

# Clean missing values
ny_nj_revenue = ny_nj_revenue.dropna()
other_revenue = other_revenue.dropna()

# Hypothesis test (Welch t-test)
t_stat, p_value = stats.ttest_ind(
    ny_nj_revenue,
    other_revenue,
    equal_var=False
)

print("p-value:", p_value)

if p_value < alpha:
    print("Reject H₀: revenue differs between regions.")
else:
    print("Fail to reject H₀: no significant difference in revenue.")
```
